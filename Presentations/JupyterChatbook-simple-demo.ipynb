## Code generation

Make a code generation LLM-persona:

```python
%%chat_meta -i cw  --prompt
@CodeWriterX|Python
```

Example generation:

```python
%%chat -i cw
Simulate a two dimensional random walk and plot it.
```

Paste the code and evaluate it:

```python
import numpy as np
import matplotlib.pyplot as plt

def random_walk_2d(steps=1000):
    directions = np.random.choice(['up', 'down', 'left', 'right'], size=steps)
    x, y = 0, 0
    x_positions = [x]
    y_positions = [y]

    for direction in directions:
        if direction == 'up':
            y += 1
        elif direction == 'down':
            y -= 1
        elif direction == 'left':
            x -= 1
        elif direction == 'right':
            x += 1
        x_positions.append(x)
        y_positions.append(y)

    return x_positions, y_positions

x, y = random_walk_2d(1000)

plt.figure(figsize=(8, 8))
plt.plot(x, y, marker='o', markersize=2, linewidth=1)
plt.title("2D Random Walk")
plt.xlabel("X position")
plt.ylabel("Y position")
plt.grid(True)
plt.axis('equal')
plt.show()
```
